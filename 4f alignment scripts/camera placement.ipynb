```python
import numpy as np
import matplotlib.pyplot as plt

from slmsuite.hardware.cameras.alliedvision import AlliedVision


CAMERA_SERIAL = "015C2"

cam = AlliedVision(
    serial=CAMERA_SERIAL,
    verbose=True,
)

print("\n=== CAMERA READY ===")
print("Shape     :", cam.shape)
print("WOI       :", cam.woi)
print("Exposure  :", cam.exposure_s, "s")
print("Bit depth :", cam.bitdepth)
```

```python
requested_exposure_s = 200e-6    # 200 us

actual_exposure_s = cam.set_exposure(
    requested_exposure_s
)

print(
    "Exposure:",
    actual_exposure_s * 1e6,
    "us"
)
```

```python
# Stop live view first
cam.live(activate=False)

#full frame ROI
roi = (0, 5496, 0, 3672)

# Use centered-ish 1600 x 1600 ROI
# roi = (1948, 1600, 1036, 1600)

#smaller ROI for testing
# roi = (2348, 800, 1436, 800)

cam.set_woi(roi)

print("ROI:", cam.woi)
print("Shape:", cam.shape)
```
